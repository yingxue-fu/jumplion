## Protein-level quantification

After precursor quantities are obtained, they are further summarized to the protein level. The precursor-level quantification results (before normalization: `precursor_quantity_summ_table.parquet`, after normalization: `precursor_quantity_summ_table_norm.parquet`) are stored in the `publication` folder.

```python
from pathlib import Path
import numpy as np
import polars as pl

prot_id = 'O14467'
prec_quantity_table = (
    pl.scan_parquet("../tests/toy_precursor_quantity_summ_table_norm.parquet")
    .filter(pl.col("Protein.Group") == prot_id)
    .collect()
)
prec_quantity_table
```

```python
prec_quality_table = (
    pl.scan_parquet("../tests/toy_precursor_quality_summ_table.parquet")
    .filter(pl.col("Protein.Group") == prot_id)
    .collect()
)
prec_quality_table
```

```python
prec_quant_level = 'ms1_ms2'

if prec_quant_level == 'ms1':
    message = "using only MS1-level precursor quantity"
elif prec_quant_level == 'ms2':
    message = "using only MS2-level precursor quantity"
elif prec_quant_level == 'ms1_ms2':
    message = "using integrated MS1- and MS2-level precursor quantities"
print("Quantifying proteins" + message)
# Extract unique sample column names (excluding primary_id and id)
other_cols = {'Precursor.Id',
              'Protein.Group',
              'Protein.Names',
              'Genes',
              'MS_type',
              'Precursor.Id_MS_type'}
sample_cols = [col for col in prec_quantity_table.columns if col not in other_cols]
```

```python
# filtering based on average quality
min_prec_quality = 0.0

prec_quality_table = prec_quality_table.with_columns(pl.mean_horizontal(sample_cols).alias("Average"))
prec_quality_table = prec_quality_table.filter(
    (pl.col("Average") >= min_prec_quality) & (pl.col("Average").is_not_nan()))

prec_quality_table
```

```python
quantity_df_sub = prec_quantity_table
quality_df_sub = prec_quality_table

prec_id_col = "Precursor.Id_MS_type"
prec_ids = set(quantity_df_sub[prec_id_col].to_list()).intersection(
    quality_df_sub[prec_id_col].to_list()
)

prec_ids = list(prec_ids)
prec_ids
```

```python
from jumplion.summarization import median_summ, topN

quantity_df_sub = quantity_df_sub.filter(pl.col(prec_id_col).is_in(prec_ids)).sort(prec_id_col)
quality_df_sub = quality_df_sub.filter(pl.col(prec_id_col).is_in(prec_ids)).sort(prec_id_col)

# Summarization
if len(prec_ids) >= 3:
    quantity_mat = quantity_df_sub.select(sample_cols).to_numpy()
    quality_mat = quality_df_sub.select(sample_cols).to_numpy()
    
    quantity_mat = quantity_mat[~np.all(np.isnan(quantity_mat), axis=1)]
    quality_mat = quality_mat[~np.all(np.isnan(quality_mat), axis=1)]
    
    quantity_summed = median_summ(quantity_mat, topN_rep=3, rep_abundance='mean')
    quality_summed = topN(quality_mat, N=3, method="mean")
else:
    max_quality_prec = (quality_df_sub.filter(pl.col("Average") == pl.col("Average").max())
                        .select(["Precursor.Id_MS_type"])).item()
    quantity_summed = quantity_df_sub.filter(pl.col("Precursor.Id_MS_type") == max_quality_prec).select(sample_cols).to_numpy()[0]
    quality_summed = quality_df_sub.filter(pl.col("Precursor.Id_MS_type") == max_quality_prec).select(sample_cols).to_numpy()[0]

quantity_summed
```
